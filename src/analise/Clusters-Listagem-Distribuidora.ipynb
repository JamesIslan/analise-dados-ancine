```python
import sklearn
from sklearn.cluster import AgglomerativeClustering, DBSCAN, KMeans
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
```

# Importação do .csv tratado

```python
df_listagem = pd.read_csv('../../db/clean/listagem-distribuidoras.csv')
```

```python
df_listagem.head()
```

# Escolha das colunas numéricas

```python
X=df_listagem[['publico_br', 'renda_br']]
```

# Utilizando o K-Means

```python
clusters_knn = KMeans(n_clusters=2, random_state=0, n_init="auto").fit_predict(X)
sns.scatterplot(data=df_listagem, x='publico_br', y='renda_br', hue=clusters_knn)
```

## Padronização para o uso do Density Based Spacial Clustering of Applications wit noise (DBSCAN)

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
clusters = DBSCAN(eps=3, min_samples=2).fit_predict(X)
sns.scatterplot(data=df_listagem, x='publico_br', y='renda_br', hue=clusters)
```

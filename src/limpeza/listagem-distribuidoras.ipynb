# Importação dos arquivos .csv

```python
import pandas as pd

def converte_int(dado: str):
  try:
    return int(dado)
  except ValueError:
    return None

df = pd.read_csv("../../db/listagem-distribuidoras.csv", sep=";", skiprows=1).iloc[:, :12]

# Renomeia colunas para padrão camel-case
df = df.rename(columns={
  "Ano de exibição": "ano_exibicao",
  "Distribuidora": "distribuidora",
  "Número de Registro Ancine": "registro_ancine",
  "Origem da empresa distribuidora": "origem_empresa",
  "Público": "publico",
  "Público Filmes Brasileiros": "publico_br",
  "Renda (R$)": "renda",
  "Renda Filmes Brasileiros (R$)": "renda_br",
  "Títulos Exibidos": "titulos_exibidos",
  "Títulos Brasileiros Exibidos": "titulos_exibidos_br",
  "Títulos Lançados": "titulos_lancados",
  "Títulos Brasileiros Lançados": "titulos_lancados_br"
})
```

## Remoção de colunas

### Remoção da coluna `registro_ancine`

```python
df.drop(columns=['registro_ancine'], inplace=True)
```

## Tratamento de colunas

### Tratamento da coluna `ano_exibicao`

```python
df['ano_exibicao'] = df['ano_exibicao'].apply(converte_int)

df.dropna(subset=['ano_exibicao'], inplace=True)
df['ano_exibicao'] = df['ano_exibicao'].astype(int)
```

### Tratamento da coluna `publico`

```python
df['publico'] = df['publico'].str.replace('.', '')
df['publico'] = df['publico'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['publico'], inplace=True)
df['publico'] = df['publico'].astype(int)
```

### Tratamento da coluna `publico_br`

```python
df['publico_br'] = df['publico_br'].str.replace('.', '')
df['publico_br'] = df['publico_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['publico_br'], inplace=True)
df['publico_br'] = df['publico_br'].astype(int)
```

### Tratamento da coluna `renda`

```python
df['renda'] = df['renda'].str.replace('.', '')
df['renda'] = df['renda'].str.replace(',', '.')
df['renda'] = df['renda'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['renda'], inplace=True)
df['renda'] = df['renda'].astype(float)
```

### Tratamento da coluna `renda_br`

```python
df['renda_br'] = df['renda_br'].str.replace('.', '')
df['renda_br'] = df['renda_br'].str.replace(',', '.')
df['renda_br'] = df['renda_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['renda_br'], inplace=True)
df['renda_br'] = df['renda_br'].astype(float)
```

### Tratamento da coluna `titulos_exibidos`

```python
df['titulos_exibidos'] = df['titulos_exibidos'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['titulos_exibidos'], inplace=True)
df['titulos_exibidos'] = df['titulos_exibidos'].astype(int)
```

### Tratamento da coluna `titulos_exibidos_br`

```python
df['titulos_exibidos_br'] = df['titulos_exibidos_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['titulos_exibidos_br'], inplace=True)
df['titulos_exibidos_br'] = df['titulos_exibidos_br'].astype(int)
```

### Tratamento da coluna `titulos_lancados`

```python
df['titulos_lancados'] = df['titulos_lancados'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['titulos_lancados'], inplace=True)
df['titulos_lancados'] = df['titulos_lancados'].astype(int)
```

### Tratamento da coluna `titulos_lancados_br`

```python
df['titulos_lancados_br'] = df['titulos_lancados_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
df.dropna(subset=['titulos_lancados_br'], inplace=True)
df['titulos_lancados_br'] = df['titulos_lancados_br'].astype(int)
```

### Tratamento de valores da coluna `origem_empresa`

```python
df['origem_empresa'].value_counts()

# Verificando através do método .value_counts(), percebemos que há apenas três possíveis valores para esta coluna, o que elimina a necessidade de tratamento específico.
```

## Exportação de .csv

```python
# Exportando DataFrame para .csv
df.to_csv("../../db/clean/listagem-distribuidoras.csv", index=False)
```

# Carregamento dos arquivos .csv

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
pasta_csv = Path("../../db/obras-nao-publicitarias")
arquivos = sorted(pasta_csv.glob("*.csv"))
dataframes = []
for a in arquivos:
  df = pd.read_csv(a, sep=";", low_memory=False)
  dataframes.append(df)

df_original = pd.concat(dataframes, ignore_index=True, sort=False)
```

# Renomeação de colunas

```python
df = df_original.copy()
df = df.rename(columns={
  'TITULO_ORIGINAL': 'titulo_original',
  'CPB': 'cpb',
  'DATA_EMISSAO_CPB': 'data_emissao_cpb',
  'SITUACAO_OBRA': 'situacao_obra',
  'TIPO_OBRA': 'tipo_obra',
  'SUBTIPO_OBRA': 'subtipo_obra',
  'CLASSIFICACAO_OBRA': 'classificacao_obra',
  'ORGANIZACAO_TEMPORAL': 'organizacao_temporal',
  'DURACAO_TOTAL_MINUTOS': 'duracao_total_minutos',
  'QUANTIDADE_EPISODIOS': 'quantidade_episodios',
  'ANO_PRODUCAO_INICIAL': 'ano_producao_inicial',
  'ANO_PRODUCAO_FINAL': 'ano_producao_final',
  'SEGMENTO_DESTINACAO_INICIAL': 'segmento_destinacao_inicial',
  'COPRODUCAO_INTERNACIONAL': 'coproducao_internacional',
  'REQUERENTE': 'requerente',
  'CNPJ_REQUERENTE': 'cnpj_requerente',
  'UF_REQUERENTE': 'uf_requerente',
  'MUNICIPIO_REQUERENTE': 'municipio_requerente'
})
```

# Tratamento de dados

## Tratamento da coluna `classificacao_obra`

- Comum: Jornalístico, eventos esportivos, programa de auditório ancorado por apresentador ou religioso
- Brasileira constituinte...: Possui vínculo com agentes econômicos ligados a grandes grupos de telecomunicação
- Brasileira independente...: Não possui vínculo com agentes econômicos ligados a grandes grupos de telecomunicação
Portanto, remover registros cujo tipo são "Comum"

```python
df = df[df['classificacao_obra'] != 'COMUM']
```

## Tratamento da coluna `organizacao_temporal`

```python
df = df[df['organizacao_temporal'] == 'NÃO SERIADA']
```

Agora que temos apenas os registros de obras não seriadas, podemos remover a coluna `organizacao_temporal`

```python
df.drop(columns=['organizacao_temporal'], inplace=True)
```

## Tratamento da coluna `tipo_obra`

```python
df = df[df['tipo_obra'] != 'REALITY-SHOW']
```

## Tratamento da coluna `duracao_total_minutos`

```python
df['duracao_total_minutos'] = df['duracao_total_minutos'].str.replace(',', '.')
df['duracao_total_minutos'] = df['duracao_total_minutos'].astype(float)
```

```python
df['duracao_total_minutos'].head(10)
```

## Adição de coluna `cat_duracao`

```python
df['cat_duracao'] = None
df.loc[df['duracao_total_minutos'] <= 15, 'cat_duracao'] = 'CURTA METRAGEM'
df.loc[(df['duracao_total_minutos'] > 15) & (df['duracao_total_minutos'] <= 70), 'cat_duracao'] = 'MÉDIA METRAGEM'
df.loc[df['duracao_total_minutos'] > 70, 'cat_duracao'] = 'LONGA METRAGEM'
```

## Tratamento da coluna `ano_producao_inicial`

```python
df['ano_producao_inicial'] = df['ano_producao_inicial'].astype(int)
```

```python
df = df[df['ano_producao_inicial'] >= 2009]
```

## Tratamento da coluna `segmento_destinacao_inicial`

```python
df['segmento_destinacao_inicial'] = (
    df['segmento_destinacao_inicial']
      .str.strip()
      .replace({
          'COMUNICAÇÃO ELETRÔNICA DE MASSA POR ASSINATURA (TV PAGA)': 'TV PAGA',
          'RADIODIFUSÃO DE SONS E IMAGENS (TV ABERTA)': 'TV ABERTA',
          'OUTROS MERCADOS - AUDIOVISUAL EM CIRCUITO RESTRITO': 'CIRCUITO RESTRITO',
          'OUTROS MERCADOS - VÍDEO POR DEMANDA': 'VÍDEO POR DEMANDA',
          'OUTROS MERCADOS - AUDIOVISUAL EM TRANSPORTE COLETIVO': 'TRANSPORTE COLETIVO',
      }, regex=False)
)
```

# Tratamento de outliers

## Remoção de outliers na coluna `duracao_total_minutos`

### Justificativa
A base de dados de obras não publicitárias não faz boa distinção entre obras seriadas e não seriadas. Apesar de haver a coluna `organizacao_temporal` (cujos valores já filtramos anteriormente para obter apenas registros não seriados), esta parece não ser suficiente para agregar todas as obras seriadas. Tendo em vista que esta análise é voltada para a exibição de filmes, o tratamento de valores discrepantes de duração se faz nesessário antes de ser iniciada qualquer etapa de visualização de dados.

```python
sns.boxplot(x='duracao_total_minutos', data=df, orient='h')
plt.title('Boxplot - Duração das produções em minutos')
plt.xlabel('Duraçao total em minutos')
plt.show()
```

### Remoção dos registros

```python
df = df[
  ((col := df['duracao_total_minutos']) > col.quantile(0.005)) &
  (col < col.quantile(0.995))
]
```

```python
# Exportando DataFrame para .csv tratado
df.to_csv("../../db/clean/obras-nao-publicitarias.csv", index=False)
```

# Importação do arquivo .csv

```python
import pandas as pd

def converte_int(dado: str):
  try:
    return int(dado)
  except ValueError:
    return None

df = pd.read_csv("../../db/salas-de-exibicao.csv", sep=";")

# Renomeia colunas para padrão snake_case
df.columns = df.columns.str.lower()
```

```python
# Filtragem de colunas úteis
df = df[['nome_sala', 'registro_sala', 'assentos_sala', 'uf_complexo', 'situacao_sala']]
```

```python
df.head()
```

```python
# Removendo registros onde registro_sala é nulo
df = df.dropna(subset=['registro_sala'])

# Converter registro_sala para inteiro
df['registro_sala'] = df['registro_sala'].astype(int)
```

```python
# Convertendo `assentos_sala` para inteiro
df['assentos_sala'] = df['assentos_sala'].astype(int)
```

```python
df.head()
```

```python
# Exportando DataFrame para .csv tratado
df.to_csv("../../db/clean/salas-de-exibicao.csv", index=False)
```

```python
# Carregamento de arquivos .csv
import pandas as pd
from pathlib import Path
pasta_csv = Path("../../db/bilheteria/distribuidoras")
arquivos = sorted(pasta_csv.glob("*.csv"))
dataframes = []
for a in arquivos:
  df = pd.read_csv(a, sep=";", low_memory=False)
  dataframes.append(df)

df_original = pd.concat(dataframes, ignore_index=True, sort=False)
```

```python
df = df_original.copy()
df = df.rename(columns={
  'DATA_EXIBICAO': 'data_exibicao',
  'TITULO_ORIGINAL': 'titulo_original',
  'TITULO_BRASIL': 'titulo_brasil',
  'CPB_ROE': 'cpb_roe',
  'PAIS_OBRA': 'pais_obra',
  'REGISTRO_SALA': 'registro_sala',
  'NOME_SALA': 'nome_sala',
  'PUBLICO': 'publico',
  'REGISTRO_GRUPO_EXIBIDOR': 'registro_grupo_exibidor',
  'REGISTRO_EXIBIDOR': 'registro_exibidor',
  'REGISTRO_COMPLEXO': 'registro_complexo',
  'MUNICIPIO_SALA_COMPLEXO': 'municipio_sala_complexo',
  'UF_SALA_COMPLEXO': 'uf_sala_complexo',
  'RAZAO_SOCIAL_DISTRIBUIDORA': 'razao_social_distribuidora',
  'CNPJ_DISTRIBUIDORA': 'cnpj_distribuidora',
  'NR_PROTOCOLO_ENVIO': 'nr_protocolo_envio',
  'DATA_HORA_ENVIO_PROTOCOLO': 'data_hora_envio_protocolo'
})
```

# Tratamento de valores nulos

```python
# Tratamento de colunas 'titulo_original' e 'titulo_brasil'
# Pegar todos os registros que não tem 'titulo_brasil' e substituir pelo valor em 'titulo_original'
df['titulo_brasil'].fillna(df['titulo_original'], inplace=True)
```

```python
# Remoção da coluna 'titulo_original'
df.drop(columns=['titulo_original'], inplace=True)
```

```python
# Renomear 'titulo_brasil' para 'titulo'
df = df.rename(columns={'titulo_brasil': 'titulo'})
```

```python
# Forçar remoção de espaços nos títulos
df['titulo'] = df['titulo'].str.strip()
```

```python
# Forçar títulos em caixa alta
df['titulo'] = df['titulo'].str.upper()
```

```python
# Remoção de valores nulos na coluna 'cpb_roe'
df.dropna(subset=['cpb_roe'], inplace=True)
```

```python
# Remoção de valores nulos na coluna 'uf_sala_complexo'
df[df['uf_sala_complexo'].isna()]['pais_obra'].value_counts()
df.dropna(subset=['uf_sala_complexo'], inplace=True)
```

```python
# Remoção de valores nulos na coluna 'registro_sala'
df.dropna(subset=['registro_sala'], inplace=True)
```

```python
df[df['pais_obra'].isna()]
```

```python
def preenche_pais_obra(registro: str):
  if registro.startswith('E'):
    return 'ESTRANGEIRO'
  else:
    return 'BRASIL'

df['pais_obra'].fillna(preenche_pais_obra, inplace=True)
```

```python
# Adição de coluna "capacidade_sala" via cruzamento de dados

df_salas = pd.read_csv("../../db/clean/salas-de-exibicao.csv", sep=",")
cap_por_sala = df_salas.set_index('registro_sala')['assentos_sala']
df['registro_sala'] = df['registro_sala'].astype(int)
df['capacidade_sala'] = df['registro_sala'].map(cap_por_sala)
```

```python
# Remoção de outliers em `publico`
df = df[df['publico'] <= df['capacidade_sala']]
```

```python
# Exportando DataFrame para .csv tratado
df.to_csv("../../db/clean/bilheteria-distribuidoras.csv", index=False)
```

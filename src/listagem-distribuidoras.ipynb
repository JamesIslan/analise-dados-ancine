```python
import pandas as pd

def converte_int(dado: str):
  try:
    return int(dado)
  except ValueError:
    return None

dist = pd.read_csv("../db/listagem-distribuidoras.csv", sep=";", skiprows=1).iloc[:, :12]

# Renomeia colunas para padrão camel-case
dist = dist.rename(columns={
  "Ano de exibição": "ano_exibicao",
  "Distribuidora": "distribuidora",
  "Número de Registro Ancine": "registro_ancine",
  "Origem da empresa distribuidora": "origem_empresa",
  "Público": "publico",
  "Público Filmes Brasileiros": "publico_br",
  "Renda (R$)": "renda",
  "Renda Filmes Brasileiros (R$)": "renda_br",
  "Títulos Exibidos": "titulos_exibidos",
  "Títulos Brasileiros Exibidos": "titulos_exibidos_br",
  "Títulos Lançados": "titulos_lancados",
  "Títulos Brasileiros Lançados": "titulos_lancados_br"
})
```

```python
# Remoção da coluna 'registro_ancine'
dist.drop(columns=['registro_ancine'], inplace=True)
```

```python
# Tratamento da coluna 'ano_exibicao'
dist['ano_exibicao'] = dist['ano_exibicao'].apply(converte_int)

dist.dropna(subset=['ano_exibicao'], inplace=True)
dist['ano_exibicao'] = dist['ano_exibicao'].astype(int)

# display(dist['ano_exibicao'].value_counts())
display(dist.info())
```

```python
# Tratamento da coluna 'publico'
dist['publico'] = dist['publico'].str.replace('.', '')
dist['publico'] = dist['publico'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['publico'], inplace=True)
dist['publico'] = dist['publico'].astype(int)
# display(dist.info())
```

```python
# Tratamento da coluna 'publico_br'
dist['publico_br'] = dist['publico_br'].str.replace('.', '')
dist['publico_br'] = dist['publico_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['publico_br'], inplace=True)
dist['publico_br'] = dist['publico_br'].astype(int)
# display(dist.info())
```

```python
# Tratamento da coluna 'renda'
dist['renda'] = dist['renda'].str.replace('.', '')
dist['renda'] = dist['renda'].str.replace(',', '.')
dist['renda'] = dist['renda'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['renda'], inplace=True)
dist['renda'] = dist['renda'].astype(float)
# display(dist.info())
```

```python
# Tratamento da coluna 'renda_br'
dist['renda_br'] = dist['renda_br'].str.replace('.', '')
dist['renda_br'] = dist['renda_br'].str.replace(',', '.')
dist['renda_br'] = dist['renda_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['renda_br'], inplace=True)
dist['renda_br'] = dist['renda_br'].astype(float)
# display(dist.info())
```

```python
# Tratamento da coluna 'titulos_exibidos'
display(dist.info())
dist['titulos_exibidos'] = dist['titulos_exibidos'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['titulos_exibidos'], inplace=True)
dist['titulos_exibidos'] = dist['titulos_exibidos'].astype(int)
# display(dist.info())
```

```python
# Tratamento da coluna 'titulos_exibidos_br'
dist['titulos_exibidos_br'] = dist['titulos_exibidos_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['titulos_exibidos_br'], inplace=True)
dist['titulos_exibidos_br'] = dist['titulos_exibidos_br'].astype(int)
# display(dist.info())
```

```python
# Tratamento da coluna 'titulos_lancados'
dist['titulos_lancados'] = dist['titulos_lancados'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['titulos_lancados'], inplace=True)
dist['titulos_lancados'] = dist['titulos_lancados'].astype(int)
# display(dist.info())
```

```python
# Tratamento da coluna 'titulos_lancados_br'
dist['titulos_lancados_br'] = dist['titulos_lancados_br'].replace({'ND': pd.NA, '#N/D': pd.NA})
dist.dropna(subset=['titulos_lancados_br'], inplace=True)
dist['titulos_lancados_br'] = dist['titulos_lancados_br'].astype(int)
# display(dist.info())
```

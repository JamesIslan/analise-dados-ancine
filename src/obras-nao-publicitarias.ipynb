# Carregamento dos arquivos .csv

```python
import pandas as pd
from pathlib import Path
pasta_csv = Path("../db/obras-nao-publicitarias")
arquivos = sorted(pasta_csv.glob("*.csv"))
dataframes = []
for a in arquivos:
  df = pd.read_csv(a, sep=";", low_memory=False)
  dataframes.append(df)

df_original = pd.concat(dataframes, ignore_index=True, sort=False)
```

# Renomeação de colunas

```python
df = df_original.copy()
df = df.rename(columns={
  'TITULO_ORIGINAL': 'titulo_original',
  'CPB': 'cpb',
  'DATA_EMISSAO_CPB': 'data_emissao_cpb',
  'SITUACAO_OBRA': 'situacao_obra',
  'TIPO_OBRA': 'tipo_obra',
  'SUBTIPO_OBRA': 'subtipo_obra',
  'CLASSIFICACAO_OBRA': 'classificacao_obra',
  'ORGANIZACAO_TEMPORAL': 'organizacao_temporal',
  'DURACAO_TOTAL_MINUTOS': 'duracao_total_minutos',
  'QUANTIDADE_EPISODIOS': 'quantidade_episodios',
  'ANO_PRODUCAO_INICIAL': 'ano_producao_inicial',
  'ANO_PRODUCAO_FINAL': 'ano_producao_final',
  'SEGMENTO_DESTINACAO_INICIAL': 'segmento_destinacao_inicial',
  'COPRODUCAO_INTERNACIONAL': 'coproducao_internacional',
  'REQUERENTE': 'requerente',
  'CNPJ_REQUERENTE': 'cnpj_requerente',
  'UF_REQUERENTE': 'uf_requerente',
  'MUNICIPIO_REQUERENTE': 'municipio_requerente'
})
```

# Tratamento de dados

## Tratamento da coluna `classificacao_obra`

- Comum: Jornalístico, eventos esportivos, programa de auditório ancorado por apresentador ou religioso
- Brasileira constituinte...: Possui vínculo com agentes econômicos ligados a grandes grupos de telecomunicação
- Brasileira independente...: Não possui vínculo com agentes econômicos ligados a grandes grupos de telecomunicação
Portanto, remover registros cujo tipo são "Comum"

```python
df = df[df['classificacao_obra'] != 'COMUM']
```

## Tratamento da coluna `organizacao_temporal`

```python
df = df[df['organizacao_temporal'] == 'NÃO SERIADA']
```

Agora que temos apenas os registros de obras não seriadas, podemos remover a coluna `organizacao_temporal`

```python
df.drop(columns=['organizacao_temporal'], inplace=True)
```

## Tratamento da coluna `duracao_total_minutos`

```python
df['duracao_total_minutos'] = df['duracao_total_minutos'].str.replace(',', '.')
df['duracao_total_minutos'] = df['duracao_total_minutos'].astype(float)
```
